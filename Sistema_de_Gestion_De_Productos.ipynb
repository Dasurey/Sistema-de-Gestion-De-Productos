```python
# Lista principal que tendra unos productos como ejemplo.
inventario = [["Monitor Valkyrie VH2410V2 24 Pulgadas", "Televisor", 850000],
              ["Teclado Gamer Redragon Kumara K552 QWERTY", "Tecnologia", 45000],
              ["Monitor Valkyrie VH2410V2 24 Pulgadas", "Monitor", 180000]
]

# Un bucle infinito que corta con la opción 5.
ejecutando = True

while ejecutando:
    # Muestra de el menú de opciones
    print("\n--- SISTEMA DE GESTIÓN DE PRODUCTOS ---")
    print("1. Agregar producto")
    print("2. Mostrar productos")
    print("3. Buscar producto")
    print("4. Eliminar producto")
    print("5. Salir")

    opcion = input("Seleccione una opción (1-5): ").strip()

    match opcion:
        # 1. Agregar producto
        case "1":
            print("\n--- 1. AGREGAR PRODUCTO ---")

            # Validación del nombre y que no este vacío
            nombre = input("Ingrese el nombre del producto: ").strip()
            while nombre == "":
                print("Error: El nombre no puede estar vacío.")
                nombre = input("Ingrese el nombre del producto: ").strip()

            # Validación de la categoría y que no esté vacía
            categoria = input("Ingrese la categoría: ").strip()
            while categoria == "":
                print("Error: La categoría no puede estar vacía.")
                categoria = input("Ingrese la categoría: ").strip()

            # Validación del precio y que sea un número entero válido
            precio_valido = False
            precio = 0
            while not precio_valido:
                precio_input = input("Ingrese el precio (solo números enteros): ").strip()
                if precio_input.isdigit(): # Verifica si son solo números
                    precio = int(precio_input)
                    if precio > 0:
                        precio_valido = True
                    else:
                        print("Error: El precio debe ser mayor a 0.")
                else:
                    print("Error: Ingrese un número entero válido (sin letras ni puntos).")

            # Crear la sublista del nuevo producto y agregarla a la lista principal
            nuevo_producto = [nombre, categoria, precio]
            inventario.append(nuevo_producto)
            print(f"\n¡Producto '{nombre}' agregado con éxito!")

        # 2. Mostrar productos
        case "2":
            print("\n--- 2. LISTA DE PRODUCTOS ---")
            if len(inventario) == 0:
                print("El inventario está vacío.")
            else:
                # Recorremos la lista usando un for con índice para numerarlos
                for i in range(len(inventario)):
                    producto = inventario[i]
                    # producto[0] es nombre, producto[1] es categoría, producto[2] es precio
                    print(f"{i + 1}. Nombre: {producto[0]} | Categoría: {producto[1]} | Precio: ${producto[2]}")

        # 3. Buscar producto
        case "3":
            print("\n--- 3. BUSCAR PRODUCTO ---")
            encontrado = False
            busqueda = input("Ingrese el nombre del producto a buscar: ").strip().lower()

            while busqueda == "":
                print("Error: No ingresó nada para buscar.")
                busqueda = input("Ingrese el nombre del producto a buscar: ").strip().lower()

            # Recorremos el inventario buscando coincidencias
            for i in range(len(inventario)):
                producto = inventario[i]
                # Convertimos el nombre a minúsculas para que la búsqueda no sea estricta
                if busqueda in producto[0].lower():
                    print("\nResultado de la búsqueda:")
                    print(f"[{i + 1}] Nombre: {producto[0]} | Categoría: {producto[1]} | Precio: ${producto[2]}")
                    encontrado = True
        
            if not encontrado:
                print("No se encontraron productos que coincidan con la búsqueda.")

        # 4. Eliminar producto
        case "4":
            print("\n--- 4. ELIMINAR PRODUCTO ---")
            if len(inventario) == 0:
                print("No hay productos para eliminar.")
            else:
                # Primero mostramos los productos para que el usuario vea los números de los productos que quiere eliminar.
                print("Productos disponibles:")
                for i in range(len(inventario)):
                    print(f"{i + 1}. {inventario[i][0]}")

                posicion_input = input("\nIngrese el número del producto que desea eliminar: ").strip()

                # Validamos que el ingreso sea un número
                if posicion_input.isdigit():
                    indice_eliminar = int(posicion_input) - 1 # Restamos 1 porque las listas empiezan en 0

                    # Validamos que el número esté dentro del rango de la lista
                    if 0 <= indice_eliminar < len(inventario):
                        # Guardamos el nombre antes de borrarlo para avisarle al usuario
                        eliminado = inventario.pop(indice_eliminar)
                        print(f"\n¡El producto '{eliminado[0]}' ha sido eliminado correctamente!")
                    else:
                        print("Error: El número de producto no existe.")
                else:
                    print("Error: Debe ingresar un número válido.")

        # 5. Salir
        case "5":
            print("\n¡Gracias por utilizar el Sistema de Gestión De Productos! Hasta luego.")
            ejecutando = False

        # Caso control si no se elije una opción válida
        case _:
            print("Opción inválida. Por favor, elija un número del 1 al 5.")
```
